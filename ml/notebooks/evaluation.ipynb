Runs the measured benchmark and plots the results. Run from the repository root.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..' if os.path.basename(os.getcwd())=='notebooks' else '.'))
from ml.evaluation.run_evaluation import run_all
report = run_all(regenerate=False)
import pandas as pd
pd.DataFrame(report['targets'])
```

```python
import matplotlib.pyplot as plt
roc = report['detection']['roc']
plt.plot([p['fpr'] for p in roc],[p['tpr'] for p in roc]); plt.plot([0,1],[0,1],'--')
plt.xlabel('false-alarm rate'); plt.ylabel('recall'); plt.title(f"ROC (AUC {report['detection']['auc']:.3f})"); plt.show()
```
